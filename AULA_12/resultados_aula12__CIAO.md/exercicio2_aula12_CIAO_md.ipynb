```python
class SOCAllocationProblem(ElementwiseProblem):
    def __init__(self):
        # TODO 1 e 2: 2 variáveis, 2 objetivos
        super().__init__(n_var=2, n_obj=2,
                         xl=np.array([1.0, 1.0]), xu=np.array([50.0, 10.0]))

    def _evaluate(self, x, out, *args, **kwargs):
        analistas, regimento = x[0], x[1]
        f1_custo = 100.0 * analistas + 50.0 * regimento      # TODO 3
        f2_tempo = 200.0 / analistas + 50.0 / regimento      # TODO 4
        out["F"] = [f1_custo, f2_tempo]

# TODO 5
res2 = minimize(SOCAllocationProblem(), NSGA2(pop_size=30),
                termination=('n_gen', 40), seed=1)
print(res2.F)
```
